Important observation

TotalCharges should ideally be numeric because it represents the total amount charged to customers.
We need to investigate why Pandas has identified it as an object.

```python
# We want to identify whether there are empty strings or other unusual values.
print(df["TotalCharges"].unique()[:20])
```

```python
# Check for empty strings
df["TotalCharges"].str.strip().eq("").sum()
```

```python
# Display affected records
df[df["TotalCharges"].str.strip() == ""]

# We want to understand why those records have no total charges.
```

Further investigation revealed 11 records containing empty strings. All 11 records had a customer tenure of zero months, indicating newly registered customers without accumulated charges.

### Cleaning

```python
# Convert TotalCharges into a numeric column
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
# Replace missing TotalCharges with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
# Verify cleaning
print("Missing values:", df["TotalCharges"].isnull().sum())

print("Data type:", df["TotalCharges"].dtype)

print("Zero TotalCharges:", (df["TotalCharges"] == 0).sum())
```

## Data Cleaning Summary

1. **Duplicate Check:** The dataset was examined for duplicate records to ensure data consistency.

3. **Data Type Correction:** The `TotalCharges` column was converted from an object data type to a numerical data type to support mathematical operations and analysis.
4. **Column Removal:** Unnecessary columns, such as `customerID`, were removed because they do not contribute meaningful predictive information.
5. **Data Validation:** The dataset was checked again to confirm that the cleaning operations were properly applied.
